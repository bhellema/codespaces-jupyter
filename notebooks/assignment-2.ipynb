```python
print('Question 2-B')

datafile['extra'].plot(kind='density')
plt.show()

datafile.groupby('group')['extra'].plot(kind='density', legend=True)
plt.show()  
```

```python
print('Question 2-C')
print('Group by group')
print(datafile.groupby('group').describe())

print('---------------------------------------')
print('Group By group and only look at the extra statistics')
print(datafile.groupby('group')['extra'].describe())
```
