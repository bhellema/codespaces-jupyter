```python
import pandas as pd
import matplotlib.pyplot as plt
datafile=pd.read_csv ("https://raw.githubusercontent.com/Psyc3000A/Data/refs/heads/main/sleep.csv")
```

```python
print("Histogram of extra sleep")
datafile['extra'].hist()
plt.show()
```

```python
print("Density plot of extra sleep")
datafile['extra'].plot(kind='density')
plt.show()
```

```python
print("Box plot of extra sleep")
datafile.boxplot(column='extra')
plt.show()
```

```python
print("Box plot of extra sleep with means")
datafile.boxplot(column='extra', showmeans=True)
plt.show()
```

```python
print("Histogram PLOT of extra sleep")
datafile['extra'].plot(kind='hist')
plt.show()
```

```python
print("Box plot of extra sleep by group")
datafile.boxplot(column='extra', by='group', showmeans=True)
plt.show()
```

```python
print("Aggregated statistics for each group:")
print(datafile.groupby('group')['extra'].agg(['count', 'mean', 'std', 'skew']))
```

```python
print('Shows the middle value of the array of numbers')
datafile.boxplot(column='extra')
plt.show()

print('Mean (mathematical average) of the data')
datafile.boxplot(column='extra', showmeans=True)
plt.show()  
```

```python
print('Question 2-B')

datafile['extra'].plot(kind='density')
plt.show()

datafile.groupby('group')['extra'].plot(kind='density', legend=True)
plt.show()  
```

```python
print('Question 2-C')
print('Group by group')
print(datafile.groupby('group').describe())

print('---------------------------------------')
print('Group By group and only look at the extra statistics')
print(datafile.groupby('group')['extra'].describe())
```

```python
print('Question 2-D')

print('Two get two histograms on the same plot, one with a comment that does nothing in the output')
datafile['extra'].hist() #this is a comment!
datafile['extra'].hist() 
```

```python
print('Question 3')
print('For each group, tell me how many observations there are, the average extra sleep, how much the values vary, and whether the distribution is skewed.')
print(datafile.groupby('group')['extra'].agg(['count', 'mean', 'std', 'skew']))
```
